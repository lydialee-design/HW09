## 一、程式實作（測試一至測試三）

### 測試一

```python
# ==========================================
# 測試一：原始資料輸入
# ==========================================
activity_names = ("桌遊體驗", "電影欣賞")
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 1. 讀取活動名稱
board_title = activity_names[0]
movie_title = activity_names[1]

# 2. 清理字串並建立保留重複的串列 (轉小寫、移空白、依逗號拆分)
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 3. 使用集合進行去重
board_set = set(board_list)
movie_set = set(movie_list)

# 4. 集合運算比較名單
common_participants = board_set & movie_set
all_participants = board_set | movie_set
only_board = board_set - movie_set
only_movie = movie_set - board_set

# 5. 排序名單並使用切片取得前兩位預覽
sorted_all = sorted(list(all_participants))
preview_notified = sorted_all[:2]

# 6. 建立 summary 字典計算人數與應收費用
summary = {
    "桌遊人數": len(board_set),
    "電影人數": len(movie_set),
    "共同人數": len(common_participants),
    "總不重複人數": len(all_participants),
    "報名人次": len(board_set) + len(movie_set),
    "桌遊應收": len(board_set) * unit_fees[board_title],
    "電影應收": len(movie_set) * unit_fees[movie_title]
}
summary["總應收"] = summary["桌遊應收"] + summary["電影應收"]

# 7. 輸出完整摘要
print("HW09 活動報名摘要 (測試一)")
print(f"活動：{board_title}，每人 {unit_fees[board_title]:.2f} 元")
print(f"活動：{movie_title}，每人 {unit_fees[movie_title]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{sorted(list(board_set))}")
print(f"電影去重名單：{sorted(list(movie_set))}")
print(f"桌遊人數：{summary['桌遊人數']} 人")
print(f"電影人數：{summary['電影人數']} 人")
print(f"共同參加者：{sorted(list(common_participants))}")
print(f"共同人數：{summary['共同人數']} 人")
print(f"只參加桌遊者：{sorted(list(only_board))}")
print(f"只參加電影者：{sorted(list(only_movie))}")
print(f"全部參加者：{sorted_all}")
print(f"總不重複人數：{summary['總不重複人數']} 人")
print(f"報名人次：{summary['報名人次']} 人次")
print(f"通知名單前兩位：{preview_notified}")
print(f"桌遊應收：{summary['桌遊應收']:.2f} 元")
print(f"電影應收：{summary['電影應收']:.2f} 元")
print(f"總應收：{summary['總應收']:.2f} 元")
```

## 二、個人問題回答

### 問題一：測試一中，整理後的桌遊串列有 4 筆資料，為什麼桌遊費用應以 3 人計算？直接把字串拆成 list，和再轉成 set，各自保留了什麼、失去了什麼？
- **為什麼以 3 人計算**：整理後的桌遊串列為 `['amy', 'ben', 'amy', 'cora']`，裡面有兩筆 `'amy'`，是因為原始名單中的 `" Amy"` 與 `"amy"` 經過移除空白與轉小寫後其實是同一位報名者。依據規則，同一個人在同一場活動重複報名只算 1 位參加者、只收 1 次費用，因此有效參加人數為 3 人，桌遊費用為 `3 × 120 = 360.00` 元。
- **拆成 `list` 與轉成 `set` 的差異**：
  - **直接拆成 `list`**：**保留了**所有人報名的先後順序，以及同一個人重複報名的次數紀錄；但**失去了**自動排除重複項目的功能，也無法直接使用集合運算子（如交集、差集）比對兩場活動名單。
  - **再轉成 `set`**：**保留了**不重複的唯一參加者名單，並獲得快速計算交集、聯集與差集的能力；但**失去了**原本的報名順序（集合沒有索引位置），也失去了某位參加者曾經重複報名幾次的紀錄。

### 問題二：測試三中，兩場活動只有 3 位不同的人，為什麼是 6 人次、總應收仍是 600 元？請用集合關係與算式說明。
- **集合關係說明**：在測試三中，桌遊集合 $A = \{\text{'amy'}, \text{'ben'}, \text{'cora'}\}$，電影集合 $B = \{\text{'amy'}, \text{'ben'}, \text{'cora'}\}$。兩者的聯集 $A \cup B = \{\text{'amy'}, \text{'ben'}, \text{'cora'}\}$ 共有 3 個元素，代表不重複的參加者總共只有 3 位；而兩者的交集 $A \cap B$ 同樣也是這 3 人，代表這 3 位參加者同時報名了兩場活動（只參加單一場的差集 $A - B$ 與 $B - A$ 皆為空集合 `[]`）。
- **人次與金額算式**：
  - **報名人次**：計算兩場各自去重後的參加人數總和，算式為 $|A| + |B| = 3 + 3 = 6$ 人次（每位參加者跨兩場活動各計 2 人次）。
  - **總應收金額**：兩場活動分開計費，桌遊應收 $3 \times 120 = 360$ 元，電影應收 $3 \times 80 = 240$ 元，兩場相加為 $360 + 240 = 600.00$ 元。
